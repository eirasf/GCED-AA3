# Autoencoders

The autoencoder is a type of network used to learn efficient encodings of unlabeled data (what is known as unsupervised learning). It is a network that has the same size in the input as in the output, since the goal of the network is to reconstruct the input with the lowest possible loss.

If what we do is reconstruct the input, what is the point of using the network? Usually, **the network has, at its middle, a layer with fewer elements than the input data**. Therefore, when reconstructing the data from the input to the output, in that layer we will have a *compressed* version of the input, which will contain most of its information.

Therefore, we can divide an autoencoder into 3 different sections, as seen in the following figure:

![](https://drive.google.com/uc?export=view&id=1yxkKZV0J0YplQAGPGJxQ2Z80Ad6L94eu)

1. **Encoder:** it is the initial part of the network, in charge of compressing the input data.
1. **Code:** it is the output of the encoder, it contains the *compressed* version of the input data.
1. **Decoder:** it is in charge of, starting from the output of the *Encoder*, reconstructing the network.

# There is life beyond the autoencoder

There are other methods to obtain useful representations of the data without relying on reconstruction. Next you will implement a method to learn representations that verify that:
    - Two variants of the same image have the same representation
    - Two different images have different representation
To do this, for each image we will generate two variants and, using a convolutional network, we will calculate their representations. The loss function will penalize equal representations for different elements of the batch, while it will also penalize differences between the representations of the two variants of the same image. We call this method [*contrastive learning*](https://arxiv.org/abs/2002.05709). In this example we will also use an output that must bring the representations to well-differentiated clusters (this favors representations with relevant information). Implement it following these guidelines:

1. Define a convolutional $model$ similar to the encoder of an autoencoder (the input is the size of the image, the output is the representation vector)
1. Define an output layer $cluster$ that, starting from the output of model, returns an output with the same number of classes as the dataset to use (the input is the representation vector), using softmax as the output activation
1. For each training batch $X$:  *# Use a high batch size, at least 128*
    1. Modify the input images with [data_augmentation](https://docs.pytorch.org/vision/0.21/transforms.html), call it $augX_1$.
    1. Modify again the input images with another different transformation (*data_augmentation_2*), call it $augX_2$.
    1. $augX_{1comp} \leftarrow model(augX_1)$ *# Use [normalize](https://docs.pytorch.org/docs/2.11/generated/torch.nn.functional.normalize.html) on the output of your model so that the representations always have the same scale*
    1. $augX_{2comp} \leftarrow model(augX_2)$
    1. $M \leftarrow augX_{1comp} ~ augX_{2comp}^T$ *# This computes the dot product (cosine similarity) of the representation of variant 1 of each element of the batch with the representation of variant 2 of each one of the elements of the batch. M will be a similarity matrix where each element (i,j) represents the similarity of the representation of variant 1 of element i with that of variant 2 of element j*
    1. $loss_M \leftarrow torch.nn.functional.cross\_entropy(M/ \tau, torch.arange(batch\_size))$ *#The pairs (i,j) must be 1 if i==j and 0 otherwise*
        1. $\tau$ is a hyperparameter that is usually defined as 5.0
    1. $cX_{1comp} \leftarrow cluster(augX_{1comp})$ *# Another network must take each representation to a cluster*
    1. $cX_{2comp} \leftarrow cluster(augX_{2comp})$
    1. $loss_C \leftarrow cX_{1comp}(1 - cX_{1comp}) + cX_{2comp}(1 - cX_{2comp})$ *# The outputs of the cluster network must be extreme (near 0 or 1, not in the middle of the range)*
    1. $loss \leftarrow loss_M + \lambda~loss_C$
        1. $\lambda$ is a hyperparameter (you can try 0.5)
