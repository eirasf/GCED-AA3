## Self-training function

We are going to create our own self-training function. The idea is to start by learning a model for the labeled data and predicting with that model the unlabeled data. From those predictions, we will take those in which the model has the highest confidence (whose probability exceeds a `thresh` threshold) and add them to the labeled data set. We will repeat the process a predetermined number of times `train_epochs`.

The pseudocode is the following:

**self_training** *(model, x_train, y_train, unlabeled_data, thresh, train_epochs)*

1. $train\_data, train\_label \leftarrow x\_train, y\_train$
1. **From** $n = 1 .. train\_epochs$ **do**
    1. Instantiate a new *model* and train it using the variables *train\_data* and *train\_label*
    2. $y\_pred \leftarrow model(unlabeled\_data)$
    3. $y\_class, y\_value \leftarrow $ Winning class in *y_pred* with its value
    4. $train\_data, train\_label \leftarrow x\_train, y\_train$
    5. **For each element** (x_u, y_c, y_v) **of the tuple** (unlabeled_data, y_class, y_value)
        1. **If** $y\_v > thresh$ **then**
            1. We add $x\_u$ and $y\_c$ to train\_data and train\_label, respectively.
4. We return the model
